```python
import sys
import os

# 1. FIX THE PATH: Tell the notebook to look one folder up for 'src'
sys.path.append(os.path.abspath(".."))

# 2. NOW IMPORT: This will work now because of the line above
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.engine import run_causal_analysis

# 3. LOAD DATA: Use '../' to go up and find the 'data' folder
try:
    df = pd.read_csv('../data/streaming_data.csv')
    naive, causal = run_causal_analysis()
    
    print(f"✅ Success! Data Loaded: {len(df)} rows")
    print(f"Naive Lift: {naive:.2f} hours")
    print(f"Causal Lift: {causal:.2f} hours")
except Exception as e:
    print(f"❌ Error: {e}")
```

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(df[df['used_feature']==1]['loyalty_score'], label='Used Feature (Treated)', fill=True, color='red')
sns.kdeplot(df[df['used_feature']==0]['loyalty_score'], label='Did Not Use (Control)', fill=True, color='grey')
plt.title("Selection Bias Visualization: Loyal Fans over-index in Feature usage")
plt.xlabel("Loyalty Score")
plt.legend()
plt.show()
# SAVE THIS AS screenshots/bias_viz.png
```

```python
plt.figure(figsize=(8, 6))
plt.bar(['Naive (Correlational)', 'Matched (Causal)'], [naive, causal], color=['#ff6666', '#2ecc71'])
plt.axhline(y=2.0, color='black', linestyle='--', label='Ground Truth (2.0)')
plt.ylabel("Estimated Watch Time Lift (Hrs)")
plt.title("Impact Analysis: Correlation vs Causation")
plt.legend()
plt.show()
# SAVE THIS AS screenshots/final_lift.png
```
